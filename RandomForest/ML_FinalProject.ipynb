```python
import pandas as pd

# Load the food poisoning complaints dataset
food_poisoning_df = pd.read_csv("food_poisoning.csv")

# Clean the column names by stripping any extra spaces
food_poisoning_df.columns = food_poisoning_df.columns.str.strip()

food_poisoning_df.columns
```

```python

# Rename columns for consistency

# For 311 food poisoning dataset
food_poisoning_df.rename(columns={
    "Incident Zip": "ZIPCODE",  # Ensuring the ZIP code column has the same name
    "Borough": "Borough",  # Standardizing the name of the borough column
    "Latitude": "Latitude",  # Ensure geospatial columns are named the same
    "Longitude": "Longitude",
    "Location Type": "Location Type",
    "Created Date": "Created_Date",
}, inplace=True)

# Filter the dataset for Brooklyn borough, "Restaurant/Bar/Deli/Bakery" location type, 
# and ZIP codes starting with "112" (Brooklyn's ZIP code range)
filtered_data = food_poisoning_df[
    (food_poisoning_df['Borough'] == "BROOKLYN") &  # Filter for Brooklyn borough
    (food_poisoning_df['Location Type'] == "Restaurant/Bar/Deli/Bakery") &  # Filter for restaurant location type
    (food_poisoning_df['ZIPCODE'].astype(str).str.startswith('112'))  # Filter for Brooklyn ZIP codes (starting with '112')
]

# Check the filtered dataset
filtered_data.head()
```

```python
# Ensure 'Inspection Date' is in datetime format and handle invalid dates
filtered_data['Created_Date'] = pd.to_datetime(filtered_data['Created_Date'], errors='coerce')


# Apply all filters at once: Borough, Location Type, ZIPCODE, and Inspection Date between 2023 and 2025
filtered_data = filtered_data[
    (filtered_data['Created_Date'].dt.year >= 2022) &  # Filter for dates >= 2023
    (filtered_data['Created_Date'].dt.year <= 2023)  # Filter for dates <= 2025
]

# Check the filtered dataset
filtered_data.head()
```

```python
# For the food poisoning complaints dataset, keep only the relevant columns
filtered_data = filtered_data[["ZIPCODE", "Longitude", "Latitude", "Borough", "Location Type", "Created_Date"]]

# Check the cleaned datasets
print(filtered_data.head())

```

```python
# Drop rows with missing geographic data
filtered_data = filtered_data.dropna(subset=["ZIPCODE", "Longitude", "Latitude", "Borough", "Location Type","Created_Date"])

# Check again for any remaining missing values
print(filtered_data.isnull().sum())
```

```python
# Check unique values in the ZIP and Borough columns of both datasets
print("Food Poisoning Dataset - ZIP codes:", filtered_data["ZIPCODE"].unique())
print("Food Poisoning Dataset - Boroughs:", filtered_data["Borough"].unique())
```

```python
# Standardize ZIP codes and Boroughs in both datasets
filtered_data["ZIPCODE"] = filtered_data["ZIPCODE"].astype(str).str.strip().str.zfill(5)  # Ensure 5-digit ZIP codes
filtered_data
# Handle specific cases for "Unspecified" boroughs or any anomalies if needed
filtered_data["Borough"] = filtered_data["Borough"].replace({"UNSPECIFIED": "UNKNOWN"})

# Check the cleaned columns
print(filtered_data[["ZIPCODE", "Borough"]].head())
```
